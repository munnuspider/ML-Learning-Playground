```python
import pandas as pd
df = pd.read_csv("globalAirQuality.csv")
#second filtering method is just to use query methods // query = copy of data
df.query("aqi > 100 & temperature > 30")[["city","aqi"]]
aqi_val = 100
aqi_data = df.query("aqi > @aqi_val & temperature > 30")[["city","aqi"]].copy() #chained comparisons
aqi_data
```

```python
#cleaning data using pandas
lf = pd.read_csv("raw_data.csv")
lf
#NaN -> automatically detected by pandas which is the missing values we need to clean
```

```python
lf.isnull() #isna() gives the same output
```

```python
lf.isnull().sum() #no of missing values per column
```

```python
lf.dropna() #drops any rows with any empty cells 
lf.dropna(axis=1) #column drop
```

```python
lf.fillna(0) # all missing cells will be filled with 0 or replace the missing ages with the mean age
age_mean = lf["age"].mean()
cleaned_data = lf.copy()
cleaned_data["age"] = cleaned_data["age"].fillna(age_mean)
cleaned_data
```

```python
#ffill() -> forward fill where we fill info from top to bottom in empty cells and bfill() -> backward fill()
lf.ffill()
lf.bfill()
```

```python
#handle dupes
lf["country"].duplicated()
```

```python
lf2 = lf.copy()
lf.drop_duplicates(inplace=True)
lf = lf2
```
